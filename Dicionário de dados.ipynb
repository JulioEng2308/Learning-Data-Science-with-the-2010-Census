# Ciência de Dados
## Trabalhos Práticos I e II: Dicionário de Dados

<br>

Base: Censo Demográfico 2010 (Amostra), IBGE. Arquivo: `censo2010_pessoas.parquet` (uma linha por pessoa, já com o resultado da junção Pessoas + Domicílios feita pelo professor).

<i>Atenção</i>: por motivos didáticos, esta base contém alguns nulos e inconsistências propositalmente introduzidos, além dos que já ocorrem naturalmente no desenho da pesquisa do IBGE. Ver TP1, seção de Qualidade de Dados.
